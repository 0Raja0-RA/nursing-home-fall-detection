# Ringkasan hasil eksperimen

Satu tempat untuk membaca hasil. Sumbernya `ml/experiments/`, jadi tidak perlu GPU maupun dataset. Tiap kesimpulan rinci ada di README folder eksperimennya.


```python
from pathlib import Path

import pandas as pd
from IPython.display import Image, Markdown, display

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "ml" / "experiments").is_dir())
EXP = ROOT / "ml" / "experiments"
pd.set_option("display.max_colwidth", 60)
```

## Riwayat training (EXP-001 sampai EXP-010)


```python
hist = pd.read_csv(EXP / "experiments.csv", on_bad_lines="skip")
display(hist[["run_id", "model", "imgsz", "val_mAP50", "test_mAP50", "test_precision", "test_recall", "status"]])
```

## Eksperimen per folder


```python
for readme in sorted(EXP.glob("*/README.md")):
    display(Markdown(readme.read_text(encoding="utf-8")))
```

## Klasifier keypoint vs aturan sudut


```python
display(pd.read_csv(EXP / "pose_classifier_cv" / "results.csv").sort_values(["validasi", "recall"], ascending=[True, False]))
```

## Oracle crop vs gambar penuh


```python
cols = ["config", "ms_per_frame", "cls2_found", "cls2_core", "cls2_kpts", "cls1_core", "cls0_core"]
display(pd.read_csv(EXP / "exp0016_oracle_crop" / "summary.csv")[cols].sort_values("cls2_core", ascending=False))
```

## Fine-tune pose (exp0013, exp0014)


```python
POSE = ROOT / "ml" / "models" / "pose" / "exp0014_yolo11m" / "eval"
display(pd.read_csv(POSE / "per_run.csv")[["sel", "seed", "ap_test", "recall_test", "precision_test", "alarm_palsu", "kejadian"]])
display(pd.read_csv(POSE / "contrasts.csv"))
```
